```python
employees=[
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]

#1
total_num = len(employees)
total_salary = 0
highest_salary = employees[0]["salary"]
lowest_salary = employees[0]["salary"]

for employee in employees:
    total_salary += employee["salary"]
    if employee["salary"] > highest_salary:
        highest_salary = employee["salary"]
    if employee["salary"] < lowest_salary:
        lowest_salary = employee["salary"]
#2
average_salary = total_salary / total_num
print("Average salary:", average_salary)
print("Highest salary:", highest_salary)
print("Lowest salary:", lowest_salary)
#3
print("Employees with salary > 65000 and rating > 4.5:")
for emp in employees:
    if emp["salary"] > 65000 and emp["rating"] > 4.5:
        print(emp["name"])
#4
employee_dict = {}
for employee in employees:
    employee_dict[employee["id"]] = employee["name"]

print ("Employee ID and Name:", employee_dict)
#5
names = []
for employee in employees:
  names.append(employee["name"])
names.sort()
names.reverse()

print("First three employee names (sorted in reverse):")
for name in names[:3]:
  print(name)

#6
employee_ids = []

for employee in employees:
   employee_ids.append(employee["id"])

print("Employee IDs:", employee_ids)
print(f"Maximum Employee ID: {max(employee_ids)}")
print(f"Minimum Employee ID: {min(employee_ids)}")

#7
dept=set()
for employee in employees:
  dept.add(employee["department"])
print("Departments:",dept)

#8
search_id=int(input("Enter the employee ID to search:"))
found=False
for employee in employees:
  if employee["id"]==search_id:
    print("Employee Name:",employee["name"])
    print("Department:",employee["department"])
    print("Salary:",employee["salary"])
    print("Rating:",employee["rating"])
    found=True
    break
if not found:
  print("Employee not found")
```
