```python
# 1
vehicle=input("enter vehicle type:")
hours=float(input("enter the number of hours:"))

if hours <= 0:
  print("invalid parking ")
elif vehicle not in ["Car","Bike","SUV"]:
  print("invalid vehicle type")
else:
    if hours < 1:
      hours = 1

    first_2_hours = 0
    additional_hour = 0

    if vehicle == "Car":
        first_2_hours = 50
        additional_hour = 20
    elif vehicle == "Bike":
        first_2_hours = 30
        additional_hour = 10
    else:
        first_2_hours = 70
        additional_hour = 30

    fee = 0
    if hours <= 2:
        fee = first_2_hours
    else:
        fee = first_2_hours + (hours - 2) * additional_hour

    print(f"Parking fee: Rs {fee:.2f}")
```

```python
# 2nd question
n = int(input("enter the number"))

highest_show = ""
highest_occupancy = 0

total_booked = 0
total_available = 0
high_demand_count = 0

for i in range(n):
    data = input().split()

    show_name = data[0]
    total_seats = int(data[1])
    booked_seats = int(data[2])

    available_seats = total_seats - booked_seats
    occupancy = (booked_seats / total_seats) * 100


    if occupancy >= 80:
        category = "House Full"
        high_demand_count += 1
    elif occupancy >= 60:
        category = "High Demand"
        high_demand_count += 1
    elif occupancy >= 40:
        category = "Moderate"
    else:
        category = "Low Demand"

    print(f"{show_name} - Available: {available_seats} - "
          f"Occupancy: {occupancy:.2f}% - {category}")


    if occupancy > highest_occupancy:
        highest_occupancy = occupancy
        highest_show = show_name

    total_booked += booked_seats
    total_available += available_seats

print("Highest Occupancy:", highest_show)
print("Total Seats Booked:", total_booked)
print("Total Seats Available:", total_available)
print("Number of Shows with High Demand:", high_demand_count)







```
